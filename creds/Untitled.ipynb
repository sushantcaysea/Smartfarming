```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the crop recommendation dataset
df = pd.read_csv("crop_data/Crop_recommendation.csv")

# 2. Define features to plot
crop_features = ["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]

# Set the theme
sns.set_theme(style="whitegrid")

# 3. Generate KDE plots
for col in crop_features:
    plt.figure(figsize=(8, 4))
    sns.kdeplot(df[col], fill=True, color="forestgreen", alpha=0.5, linewidth=2.5)
    plt.title(f"Density Distribution of {col.capitalize()} in Soil/Climate", fontsize=14, fontweight="bold", pad=15)
    plt.xlabel(col.capitalize(), fontsize=12)
    plt.ylabel("Density", fontsize=12)
    plt.tight_layout()
    plt.show()
```
